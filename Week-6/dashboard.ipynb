```python
# WEEK 6: DATA VISUALIZATION WITH SEABORN (No Plotly)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("sales_data.csv")  # Basic sales dataset


plt.figure(figsize=(8,5))
sns.scatterplot(x='Quantity', y='Price', data=df, color='purple', s=80)
plt.title("Quantity vs Price")
plt.xlabel("Quantity")
plt.ylabel("Price")
plt.show()


plt.figure(figsize=(8,5))
sns.boxplot(x='Product', y='Price', data=df, palette='Set2')
plt.title("Price Distribution by Product")
plt.show()

plt.figure(figsize=(8,5))
sns.violinplot(x='Product', y='Price', data=df, palette='Set3')
plt.title("Price Distribution (Violin Plot) by Product")
plt.show()


numeric_cols = ['Quantity', 'Price', 'Total_Sales']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(6,5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()

pivot_table = pd.pivot_table(df, values='Total_Sales', index='Product', columns='Region', aggfunc='sum', fill_value=0)
plt.figure(figsize=(10,6))
sns.heatmap(pivot_table, annot=True, fmt="d", cmap="Blues")
plt.title("Revenue by Product & Region")
plt.show()


fig, axes = plt.subplots(2, 2, figsize=(12,10))
sns.histplot(df['Price'], ax=axes[0,0], color='skyblue')
axes[0,0].set_title("Price Distribution")

sns.boxplot(x='Product', y='Price', data=df, ax=axes[0,1], palette='Set2')
axes[0,1].set_title("Price by Product")

sns.scatterplot(x='Quantity', y='Total_Sales', data=df, ax=axes[1,0], color='green')
axes[1,0].set_title("Quantity vs Total Sales")

sns.barplot(x='Region', y='Total_Sales', data=df, ax=axes[1,1], palette='Set1')
axes[1,1].set_title("Revenue by Region")

plt.tight_layout()
plt.show()


plt.figure(figsize=(8,5))
scatter = plt.scatter(df['Quantity'], df['Total_Sales'], c=df['Price'], cmap='viridis', s=80)
plt.colorbar(scatter, label='Price')
plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.title("Quantity vs Total Sales (Colored by Price)")
plt.show()

```
